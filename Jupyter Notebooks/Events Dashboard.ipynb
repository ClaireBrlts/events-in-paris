```python
import pandas as pd
import streamlit as st
```

```python
url1 = 'events.csv'
url2 = 'daily_meteo.csv'
url3 = 'travel_times.csv'
```

```python
events = pd.read_csv(url1, dtype={8: str})
meteo = pd.read_csv(url2)
time = pd.read_csv(url3)
```

```python
merged_events_time = pd.merge(events, time, on='event_id', how='left')
```

```python
events_final = pd.merge(merged_events_time, meteo, on='date', how= 'left')
```

```python
events_final.columns
```

```python
events_final = events_final.drop(['latitude_x', 'longitude_x'], axis=1)
```

```python
events_final = events_final.rename(columns={
    'latitude_y': 'latitude',
    'longitude_y': 'longitude'
})
```

```python
events_final['date'] = pd.to_datetime(events_final['date'])
```

```python
events_final = events_final.dropna(subset=['latitude', 'longitude'])
```

```python
events_final.to_csv("events_final_streamlit_dash.csv", index=False, encoding='utf-8')
```

```python
events_final.dtypes
```

```python
# 1. Identify the theme columns
theme_cols = [col for col in events_final.columns if col.startswith('theme')]

# 2. Melt the dataframe: convert wide format (many date columns) into long format (single column)
events_final_melted = events_final.melt(
    id_vars=[col for col in events.columns if col not in theme_cols],
    value_vars=theme_cols,
    var_name='theme_number',
    value_name='theme'
)

# 3. Drop rows where the date is missing (optional but often useful)
events_final_melted = events_final_melted.dropna(subset=['theme'])

```
